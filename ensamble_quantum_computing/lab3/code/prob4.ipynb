```python
import numpy as np
from qutip import Qobj, basis, tensor, identity, sigmax, sigmaz

# Define single-qubit identity and Pauli operators
I = identity(2)
X = sigmax()
Z = sigmaz()
```

```python
# Construct the CNOT using the product-operator expansion
# CNOT_{(1->2)} = (1/2)*( I⊗I + Z⊗I + I⊗X - Z⊗X )
CNOT_12 = 0.5 * (tensor(I, I) 
                 + tensor(Z, I) 
                 + tensor(I, X) 
                 - tensor(Z, X))

# Print or inspect the resulting operator
print("CNOT gate (control qubit = 1, target qubit = 2):")
print(CNOT_12)
```

```python
ket0 = basis(2, 0)  # |0>
ket1 = basis(2, 1)  # |1>

states = {
    "|00>": tensor(ket0, ket0),
    "|01>": tensor(ket0, ket1),
    "|10>": tensor(ket1, ket0),
    "|11>": tensor(ket1, ket1),
}

print("Applying CNOT to computational basis states:")
for label, state in states.items():
    out_state = CNOT_12 * state
    print(f"CNOT |{label}> =")
    print(out_state)
    print()

superpos = (states["|00>"] + states["|10>"]).unit()
out_superpos = CNOT_12 * superpos
print("Superposition example:")
print("Initial state: (|00> + |10>) / sqrt(2)")
print("Result after CNOT:")
print(out_superpos)
```
