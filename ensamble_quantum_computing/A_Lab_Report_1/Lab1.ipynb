7. **The number of points and the dwell time**: For this experiment, we used 2048 number of points and 500 $\mu\text{s}$ dwell time. The time interval between successive digitized points in the FID. It is inversely related to the spectral width (SW), SW=1/dwell_time. The total number of data points sampled during the acquisition. The total acquisition time (AQ) is therefore AQ = N * dwell time. In the time domain, the initial amplitude of the FID (right after the RF pulse) is essentially independent of how frequently you sample it, provided you do sample fast enough to avoid digitization artifacts. In the frequency domain, the peak amplitude can appear to change if the software scales the Fourier transform differently for different N. However, with proper normalization, the true signal amplitude (or integrated peak area) does not fundamentally change just by changing dwell time or number of points (assuming the same total time and no aliasing). With a fixed AQ, Shorter dwell time, more points → larger spectral width but the same overall time-domain length → same resolution and similar SNR; Longer dwell time, fewer points → smaller spectral width but still the same AQ → same resolution, similar SNR.

In our case our peak resoltion is 1.1Hz, our window is 32ppm.

## Experiment 3: Inversion Recovery

### Concepts and Intuitions:

1. $T_1$ time: $T_1$ time is the time for the net magnetization return to $(1-\frac{1}{e})$ of it's equilibrium after being perturbed by RF pulses. This relaxation process occurs due to interactions between the nuclear spins and their surrounding molecular environment (the lattice), resulting in the transfer of energy from the spins to the lattice. As a result, $T_1$ reflects the efficiency of energy dissipation from the spin system to its surroundings.
2. The mathematical expression for $T_1$ is $I_t = I_0(1-e^{\frac{t}{T_1}})$

## Experiment 4: Saturation Recovery

### Concepts and Intuitions:

1. How saturation recovery works: 
    Saturation recovery works in the following steps:
    1). Wait for the spin to relax

    2). Rotate the spin to the transvers plane and wait for $T_2*$ time, where it is effected by the dephasing caused by slightly inhomogenouse magnetic field.

    Now the spin is spreaded on the transvers plane, let's say x-y plane

    3). Apply a $\pi/2$ pulse, let's assume along the x axis, any axis in the transvers plane is also good, to the spin and wait the same amount of time as last time. The inhomogenous magnetic field intorduce dephasing along x-y plane again. Now the spin is spreaded all over the place, the total magnetic moment is averaged out to be 0, shown in Figure 1 and 2 below.

    Figure 1. a messy bloch sphere
    
    
        

    Figure 2. the average of all those vectors

    

    If we apply a $\pi/2$ pulse right after these procedure, it is easy to see that the receiver will get nothing.

    4). Now that the vector is spreaded all over the place, we wait a bit so that the vectors can return to the equilibrium. If we wait a short period of time, this is what the vectors and there sum are going to look like:
    
    Figure 3. half-way towards the equilibrium

    

    Figure 4. their sum

    

    So it is really easy to deduct what will happen if we apply a $\pi/2$ pulse at this point: we get a not so strong signal. If we wait longer, the signal will get stronger, until the spins are allowed to return to the equilibrium, at that point we will get the max signal.

    5). It is pretty clear now then, we do this experiment many times with different waiting duration, then fit it to the $T_1$ curve mentioned in experiment three, the $T_1$ is thereby obtained.

Please note that here in step 2 and 3 I used a simplified discription to saturate the bloch sphere, in practice we need to repeat step 3 more times so the saturation is good enough.

2. Why $T_2^*$ between each $\pi/2$ pulses when creating a saturation? To maximize the effect of the saturation in the transvers plane.

1. How polarization transfer increases the SNR of ${}^{13}C$ signal by taking NOE into accout?
    
    To understand polarization transfer, let's look at how it is conducted: 1). the proton is excited, 2). proton and carbon are allowed to exchange magnetization with J-coupling, 3). a sequence of $\pi$ pulses are applied to allow transverse proton coherence transferred to a transverse carbon coherence. 

    To understand NOE, we first look at this Solomon equation:

    $$ \frac{d(I_Z-I_Z^0)}{dt} = -R_I(I_Z - I_Z^0) - \sigma_{IS}(S_Z - S_Z^0)$$

      if S spin magnetization deviates from equilirium, the change caused by this in the I spin magnetization is proportional to such deviation. This change in I spin magnetization is effectively the change of the intensity in the corresponding spectrum.

      By combining the two, the effect of 13C FID is significantly better.
