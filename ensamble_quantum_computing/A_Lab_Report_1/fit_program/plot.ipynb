```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# --- Parameters ---
n_vectors = 100  # Number of random vectors to plot

# --- Create a 3D figure ---
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# --- Plot the Bloch sphere (wireframe) ---
# Generate data for the sphere using spherical coordinates
u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
x_sphere = np.outer(np.cos(u), np.sin(v))
y_sphere = np.outer(np.sin(u), np.sin(v))
z_sphere = np.outer(np.ones(np.size(u)), np.cos(v))

# Plot the wireframe of the sphere
ax.plot_wireframe(x_sphere, y_sphere, z_sphere, color='gray', alpha=0.5)

# --- Generate random vectors on the surface of the sphere ---
# Uniformly distributed random directions:
#   phi   in [0, 2π)
#   theta in [0, π] (polar angle)
phi = 2 * np.pi * np.random.rand(n_vectors) * 0.6
theta = np.arccos(2 * np.random.rand(n_vectors) - 1) * 0.6

# Convert spherical coordinates to Cartesian
x_rand = np.sin(theta) * np.cos(phi)
y_rand = np.sin(theta) * np.sin(phi)
z_rand = np.cos(theta)

# --- Plot the random vectors ---
for i in range(n_vectors):
    ax.quiver(0, 0, 0, 
              x_rand[i], y_rand[i], z_rand[i], 
              length=1.0, color='red', arrow_length_ratio=0.1)

# --- Adjust the axes ---
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-1, 1])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Bloch Sphere with Random Vectors')
plt.show()
# --- Show the plot ---

# --- Figure 2 ---

fig2 = plt.figure(figsize=(6, 6))
ax2 = fig2.add_subplot(111, projection='3d')

u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
x_sphere = np.outer(np.cos(u), np.sin(v))
y_sphere = np.outer(np.sin(u), np.sin(v))
z_sphere = np.outer(np.ones(np.size(u)), np.cos(v))

# Plot the wireframe of the sphere
ax2.plot_wireframe(x_sphere, y_sphere, z_sphere, color='gray', alpha=0.5)

# Compute the vector sum
sum_x = np.sum(x_rand)/n_vectors
sum_y = np.sum(y_rand)/n_vectors
sum_z = np.sum(z_rand)/n_vectors

# Plot the sum vector
ax2.quiver(0, 0, 0,
           sum_x, sum_y, sum_z,
           color='blue', arrow_length_ratio=0.1)

# Optionally, set axis limits based on the maximum expected range
# For random vectors on a unit sphere, the sum might range up to n_vectors in magnitude.
max_range = n_vectors
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-1, 1])

ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')
ax2.set_title('Sum of Random Vectors')

plt.show()
```
