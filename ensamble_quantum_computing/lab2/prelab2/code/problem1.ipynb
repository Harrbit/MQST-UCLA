```python
from matplotlib import pyplot as plt
import numpy as np

T_2 = 10
Omega = 5
omega = np.linspace(0, 10, 1000)

phi_list = np.linspace(0, 2*np.pi, 5)
for phi in phi_list:
    R = (T_2*np.cos(phi) - T_2**2 * (omega - Omega)*np.sin(phi))/(1 + T_2**2 * (omega - Omega)**2)
    I = (T_2*np.sin(phi) + T_2**2 * (omega - Omega)*np.cos(phi))/(1 + T_2**2 * (omega - Omega)**2)
    if phi == 0:
        plt.plot(omega, R, 'x', label=f"R at phi={phi:.2f}")
        # plt.plot(omega, I,'x', label=f"I at phi={phi:.2f}")
    else:
        plt.plot(omega, R, label=f"R at phi={phi:.2f}")
        # plt.plot(omega, I, label=f"I at phi={phi:.2f}")
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(len(phi_list), 1, figsize=(10, 20))

for i, phi in enumerate(phi_list):
    R = (T_2*np.cos(phi) - T_2**2 * (omega - Omega)*np.sin(phi))/(1 + T_2**2 * (omega - Omega)**2)
    I = (T_2*np.sin(phi) + T_2**2 * (omega - Omega)*np.cos(phi))/(1 + T_2**2 * (omega - Omega)**2)
    axs[i].plot(omega, R, label=f"R at phi={phi/np.pi:.1f}π")
    axs[i].plot(omega, I, label=f"I at phi={phi/np.pi:.1f}π")
    axs[i].legend()
    axs[i].set_title(f"phi={phi/np.pi:.1f}π")

plt.tight_layout()
plt.show()
```
